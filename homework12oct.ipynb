```python

db = {
    1: {"Maths": 78, "Physics": 95, "Chemistry": 85},
    2: {"Maths": 97, "Physics": 94, "Chemistry": 92},
    3: {"Maths": 78, "Physics": 90, "Chemistry": 98}
}

subjects = ["Maths", "Physics", "Chemistry"]

for subject in subjects:
    topper_roll = 0
    topper_marks = 0

    for roll, marks in db.items():
        if marks[subject] > topper_marks:
            topper_roll = roll
            topper_marks = marks[subject]

    print(f"{subject} topper is Roll No. {topper_roll} with marks {topper_marks}")
```
